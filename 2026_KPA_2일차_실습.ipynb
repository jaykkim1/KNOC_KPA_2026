```python
import gradio as gr
import requests

def get_weather(city_name):
    # 1. 도시 이름을 위도/경도로 변환 (Open-Meteo Geocoding API)
    geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=1&language=en&format=json"

    try:
        geo_res = requests.get(geo_url).json()
        if "results" not in geo_res or not geo_res["results"]:
            return f"❌ '{city_name}'에 해당하는 도시를 찾을 수 없습니다. 올바른 영문 도시명(예: Seoul, New York, Tokyo)을 입력해주세요."

        location = geo_res["results"][0]
        lat = location["latitude"]
        lon = location["longitude"]
        country = location.get("country", "")
        name = location.get("name", city_name)

        # 2. 해당 좌표의 실시간 날씨 정보 조회 (Open-Meteo Weather API)
        weather_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current=temperature_2m,relative_humidity_2m,apparent_temperature,wind_speed_10m"
        weather_res = requests.get(weather_url).json()

        current = weather_res["current"]
        temp = current["temperature_2m"]
        apparent_temp = current["apparent_temperature"]
        humidity = current["relative_humidity_2m"]
        wind_speed = current["wind_speed_10m"]

        # 3. 보기 좋게 결과 포맷팅
        result_text = f"""
        🌍 **위치:** {name}, {country} (위도: {lat}, 경도: {lon})

        🌡️️ **현재 기온:** {temp} °C
         відчутие (체감 기온): {apparent_temp} °C
         💧 **습도:** {humidity} %
         🌬️ **풍속:** {wind_speed} m/s
        """
        return result_text

    except Exception as e:
        return f"⚠️ 데이터를 불러오는 중 오류가 발생했습니다: {str(e)}"

# Gradio 인터페이스 구성
demo = gr.Interface(
    fn=get_weather,
    inputs=gr.Textbox(label="도시 이름 (영문)", placeholder="예: Seoul, London, New York", value="Seoul"),
    outputs=gr.Markdown(label="실시간 날씨 정보"),
    title="🌦️ 실시간 날씨 확인 웹앱",
    description="도시 이름을 영문으로 입력하면 Open-Meteo API를 통해 실시간 날씨 정보를 가져옵니다."
)

# 코랩에서 실행할 때는 share=True를 주어야 외부 접속 가능한 링크가 생성됩니다.
demo.launch(share=True, debug=True)
```

```python
import gradio as gr
import requests

def get_weather(city_name):
    if not city_name.strip():
        return "❌ 도시 이름을 입력해주세요."

    # 1. 도시 이름을 위도/경도로 변환 (language=ko를 지정하여 한글 검색 지원)
    geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=1&language=ko&format=json"

    try:
        geo_res = requests.get(geo_url).json()
        if "results" not in geo_res or not geo_res["results"]:
            return f"❌ '{city_name}'에 해당하는 도시를 찾을 수 없습니다. 올바른 도시명(예: 서울, 도쿄, 뉴욕)을 입력해주세요."

        location = geo_res["results"][0]
        lat = location["latitude"]
        lon = location["longitude"]
        country = location.get("country", "")
        name = location.get("name", city_name)

        # 2. 해당 좌표의 실시간 날씨 정보 조회
        weather_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current=temperature_2m,relative_humidity_2m,apparent_temperature,wind_speed_10m"
        weather_res = requests.get(weather_url).json()

        current = weather_res["current"]
        temp = current["temperature_2m"]
        apparent_temp = current["apparent_temperature"]
        humidity = current["relative_humidity_2m"]
        wind_speed = current["wind_speed_10m"]

        # 3. 결과 포맷팅
        result_text = f"""
        ### 🌍 위치: {name} ({country})
        * **위도 / 경도:** {lat}, {lon}
        * **🌡️️ 현재 기온:** {temp} °C
        * **🧥 체감 기온:** {apparent_temp} °C
        * **💧 습도:** {humidity} %
        * **🌬️ 풍속:** {wind_speed} m/s
        """
        return result_text

    except Exception as e:
        return f"⚠️ 데이터를 불러오는 중 오류가 발생했습니다: {str(e)}"

# Gradio 인터페이스 구성
demo = gr.Interface(
    fn=get_weather,
    inputs=gr.Textbox(label="도시 이름", placeholder="예: 서울, 부산, 도쿄, 뉴욕", value="서울"),
    outputs=gr.Markdown(label="실시간 날씨 정보"),
    title="🌦️ 실시간 날씨 확인 웹앱 (한글 지원)",
    description="도시 이름을 한글(또는 영문)로 입력하면 실시간 날씨 정보를 가져옵니다."
)

# 코랩 실행용
demo.launch(share=True, debug=True)
```

```python
import random

def play_ladder():
    print("=" * 40)
    print("🎲 사다리 게임을 시작합니다! 🎲")
    print("=" * 40)

    # 1. 참가자 입력 받기
    p_input = input("참가자 이름을 쉼표(,)로 구분해서 입력해주세요\n(예: 민수,영희,철수,지영): ")
    participants = [p.strip() for p in p_input.split(",") if p.strip()]

    if not participants:
        print("❌ 참가자가 없습니다. 다시 실행해주세요.")
        return

    # 2. 결과(당첨 내용) 입력 받기
    r_input = input(f"\n결과 항목을 쉼표(,)로 구분해서 입력해주세요\n(참가자 수와 동일하게 맞춰주세요. 예: 아이패드,커피,꽝,꽝): ")
    results = [r.strip() for r in r_input.split(",") if r.strip()]

    # 참가자와 결과 수 맞추기 보정
    if len(results) < len(participants):
        print(f"⚠️ 결과 개수가 부족하여 부족한 만큼 '꽝'으로 채웁니다.")
        results.extend(["꽝"] * (len(participants) - len(results)))
    elif len(results) > len(participants):
        results = results[:len(participants)]

    num = len(participants)

    # 3. 사다리 구조 생성 (가로 선 만들기)
    # 층수(높이)는 참가자 수의 2배 정도로 설정
    height = max(6, num * 2)
    # 다리 연결 상태를 저장할 2차원 리스트 (True면 오른쪽 가로선 있음)
    ladders = [[False for _ in range(num - 1)] for _ in range(height)]

    for h in range(height):
        i = 0
        while i < num - 1:
            # 바로 왼쪽에 다리가 없으면 50% 확률로 다리 생성
            if i == 0 or not ladders[h][i-1]:
                if random.random() < 0.45:  # 다리가 생길 확률 (45%)
                    ladders[h][i] = True
            i += 1

    # 4. 사다리 시각화 그리기
    print("\n[사다리 도면]")
    # 윗줄 참가자 이름 출력 (길이에 맞춰 정렬)
    name_line = "  ".join([f"{p:^6}" for p in participants])
    print(name_line)

    # 사다리 몸통 그리기
    for h in range(height):
        row_str = "    "
        for i in range(num - 1):
            if ladders[h][i]:
                row_str += "|------"
            else:
                row_str += "|      "
        row_str += "|"
        print(row_str)

    # 아랫줄 결과 출력
    result_line = "  ".join([f"{r:^6}" for r in results])
    print(result_line)
    print("=" * 40)

    # 5. 각 참가자별 사다리 추적 및 결과 계산
    print("\n📊 [최종 결과]")
    final_mapping = {}

    for start_idx, name in enumerate(participants):
        current_pos = start_idx
        for h in range(height):
            # 왼쪽으로 가는 다리가 있으면 이동
            if current_pos > 0 and ladders[h][current_pos - 1]:
                current_pos -= 1
            # 오른쪽으로 가는 다리가 있으면 이동
            elif current_pos < num - 1 and ladders[h][current_pos]:
                current_pos += 1

        assigned_result = results[current_pos]
        final_mapping[name] = assigned_result
        print(f"👉 {name} ➔ {assigned_result}")

# 함수 실행
play_ladder()
```

```python
import random

def play_ladder():
    print("=" * 50)
    print("🎲 직관적인 사다리 타기 게임 🎲")
    print("=" * 50)

    # 1. 입력 받기
    p_input = input("참가자 이름을 쉼표(,)로 구분해서 입력해주세요\n(예: 철수,영희,민수,지원): ")
    participants = [p.strip() for p in p_input.split(",") if p.strip()]

    if not participants:
        print("❌ 참가자가 없습니다. 다시 실행해주세요.")
        return

    r_input = input(f"\n결과 항목을 쉼표(,)로 구분해서 입력해주세요\n(참가자 수와 동일하게 맞춰주세요. 예: 아이패드,커피,꽝,꽝): ")
    results = [r.strip() for r in r_input.split(",") if r.strip()]

    # 개수 맞추기 보정
    if len(results) < len(participants):
        results.extend(["꽝"] * (len(participants) - len(results)))
    elif len(results) > len(participants):
        results = results[:len(participants)]

    num = len(participants)

    # 2. 사다리 구조 생성 (높이 7층)
    height = 7
    ladders = [[False for _ in range(num - 1)] for _ in range(height)]

    for h in range(height):
        i = 0
        while i < num - 1:
            if i == 0 or not ladders[h][i-1]:
                if random.random() < 0.5:  # 다리 생성 확률 50%
                    ladders[h][i] = True
            i += 1

    # 3. 직관적인 사다리 도면 그리기
    print("\n" + "=" * 10 + " [사다리 도면] " + "=" * 10)

    # 이름 출력 (너비 고정 정렬)
    name_str = "    "
    for p in participants:
        name_str += f"{p:^8}"
    print(name_str)

    # 이름과 사다리 연결하는 시작선
    start_line = "    "
    for _ in range(num):
        start_line += f"{'│':^8}"
    print(start_line)

    # 사다리 몸통 그리기
    for h in range(height):
        row_str = "    "
        for i in range(num):
            # 세로줄 그리기
            row_str += "│"

            # 오른쪽으로 가는 다리가 있으면 ─── 그리기, 없으면 공백
            if i < num - 1:
                if ladders[h][i]:
                    row_str += "──────"
                else:
                    row_str += "      "
        print(row_str)

    # 결과와 사다리 연결하는 끝선
    end_line = "    "
    for _ in range(num):
        end_line += f"{'│':^8}"
    print(end_line)

    # 결과 출력 (너비 고정 정렬)
    result_str = "    "
    for r in results:
        result_str += f"{r:^8}"
    print(result_str)
    print("=" * 35)

    # 4. 사다리 경로 추적 및 결과 계산
    print("\n📊 [최종 결과]")
    final_mapping = {}

    for start_idx, name in enumerate(participants):
        current_pos = start_idx
        for h in range(height):
            if current_pos > 0 and ladders[h][current_pos - 1]:
                current_pos -= 1
            elif current_pos < num - 1 and ladders[h][current_pos]:
                current_pos += 1

        assigned_result = results[current_pos]
        final_mapping[name] = assigned_result
        print(f"👉 {name} ➔ {assigned_result}")

# 함수 실행
play_ladder()
```
