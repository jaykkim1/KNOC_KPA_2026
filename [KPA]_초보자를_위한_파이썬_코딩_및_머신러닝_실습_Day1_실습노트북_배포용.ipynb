## 데이터 처리/연산 툴: Numpy(넘파이)  (슬라이드 36~38)

````python
import numpy as np

a = np.array([1, 2, 3, 4])        # 1차원 배열
b = np.array([[1, 2], [3, 4]])    # 2차원 배열(행렬)
print(a)
print(b)

print(np.shape(a))  # (4,)
print(np.shape(b))  # (2,2)
````

````python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(x + y)     # 원소별 덧셈
print(x * y)     # 원소별 곱셈
print(x.dot(y))  # 내적

data = np.array([1, 2, 3, 4, 5])
print(np.mean(data))
print(np.std(data))
````

### 개념: 리스트와 넘파이 연산 차이  *(참고: 슬라이드 38)*
리스트는 그대로 연산이 안 되고(`*2`를 하면 반복됨), 넘파이 배열은 원소별로 계산됩니다.

````python
test_list = [1, 2, 3]
test_np = np.array(test_list)

print(test_list * 2)  # [1, 2, 3, 1, 2, 3] ← 리스트는 반복
print(test_np * 2)    # [2 4 6]           ← 넘파이는 원소별 연산
````

## [실습] 조건별로 데이터 나눠서 각각 저장하기  (슬라이드 50)
부서별로 데이터를 나누어 부서마다 별도의 엑셀 파일로 저장해봅니다.

````python
departments = df["부서"].unique()

for dept in departments:
    dept_df = df[df["부서"] == dept]
    dept_df.to_excel(f"{dept}팀_명단.xlsx", index=False)

print("저장 완료:", [f"{d}팀_명단.xlsx" for d in departments])
````

## [실습] 정리한 데이터 새 파일로 저장 & 서식 입히기  (슬라이드 52)

````python
from openpyxl import load_workbook
from openpyxl.styles import Font

df.to_excel("정리결과.xlsx", index=False)

wb = load_workbook("정리결과.xlsx") # wb: 엑셀파일 전체
ws = wb.active                      # ws: 엑셀파일 내 활성화된 시트만
for cell in ws[2]:                  # ws내에서 2번째 줄에 대해 각 셀들을 순차적으로 반환
    cell.font = Font(bold=True)
wb.save("정리결과.xlsx")
print("저장 완료: 정리결과.xlsx")
````

- **shutil** → 파일을 옮기거나 복사하기


## [실습] 폴더 정리 스크립트 만들기 — 시나리오①  (슬라이드 60)
`자동화_원본파일` 폴더 안에 팀 이름이 다르게 섞인 파일들을 팀별 폴더로 자동 이동합니다.

````python
import glob, shutil, os

source_dir = "자동화_원본파일"

for f in glob.glob(os.path.join(source_dir, "*.xlsx")):
    fname = os.path.basename(f)
    team = fname.split("_")[0]              # 파일명 앞부분(팀 이름) 추출
    target_dir = os.path.join(source_dir, team)
    os.makedirs(target_dir, exist_ok=True)
    shutil.move(f, os.path.join(target_dir, fname))

print("정리 완료:", os.listdir(source_dir))
````

## [실습] 여러 파일 병합 실습 — 시나리오②  (슬라이드 62)
`실적` 폴더의 12개월 파일을 하나로 병합하고 "월" 열을 추가합니다.

````python
import glob, pandas as pd, os

files = glob.glob("실적/*.xlsx")
dfs = []
for f in files:
    d = pd.read_excel(f)
    month = os.path.basename(f).replace("실적_", "").replace(".xlsx", "")
    d["월"] = month
    dfs.append(d)

all_df = pd.concat(dfs, ignore_index=True)
print(all_df.shape)
all_df.head()
````
